```python
df3.select_dtypes(include=['object']).columns
```

```python
pd.concat([df['MSZoning'].value_counts()/df.shape[0]*100,
           df3['MSZoning'].value_counts()/df3.shape[0]*100],axis=1,
              keys=["MSZoning_original","MSZoning_clean"])
```

```python
def cat_var_dist(var):
    return pd.concat([df['MSZoning'].value_counts()/df.shape[0]*100,
           df3['MSZoning'].value_counts()/df3.shape[0]*100],axis=1,
              keys=[var+"_original",var+"_clean"])
```

```python
cat_var_dist("MSZoning")
```

```python
sm=df3.select_dtypes(include=['object']).columns
sm

for i in sm:
    print(cat_var_dist(i))
    print()
```
