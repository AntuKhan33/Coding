```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option("display.max_rows",None)
pd.set_option("display.max_columns",None)
```

```python
df = pd.read_csv("train.csv")
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
null_var=df.isnull().sum()/df.shape[0]*100
null_var
```

```python
drop_columns=null_var[null_var>20].keys()
drop_columns
```

```python
df2=df.drop(columns=drop_columns,axis=1)
```

```python
df2.head()
```

```python
df2.shape
```

```python
df3=df2.select_dtypes(include=['int64','float64'])
df3.head()
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(df3.isnull())
```

```python
df3[df3.isnull().any(axis=1)]
```

```python
df3.isnull().sum()
```

```python
missing_num_var=[var for var in df3.columns if df3[var].isnull().sum()>0]
missing_num_var
```

```python
plt.figure(figsize=(10,10))
sns.set()

for i ,var in enumerate(missing_num_var):
    plt.subplot(3,3,i+1)
    sns.distplot(df3[var],bins=20)
```

```python
df4=df3.fillna(df3.mean())
df4.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,20))
sns.set()

for i ,var in enumerate(missing_num_var):
    plt.subplot(3,3,i+1)
    sns.distplot(df3[var],bins=20,label="Original",kde_kws={'linewidth':8, 'color':'red'})
    sns.distplot(df4[var],bins=20,label="Mean",kde_kws={'linewidth':5,"color":"Green"})
    plt.legend()
```

```python
df5=df3.fillna(df.median())
df5.isnull().sum().sum()
```

```python
plt.figure(figsize=(10,10))
sns.set()
for i,var in enumerate(missing_num_var):
    plt.subplot(2,2,i+1)
    sns.distplot(df3[var],bins=20,hist=False,kde_kws={'linewidth':8,"color":'red'},label="Original")
    sns.distplot(df4[var],bins=20,hist=False,kde_kws={'linewidth':5,"color":"green"},label="Mean")
    sns.distplot(df5[var],bins=20,hist=False,kde_kws={'linewidth':3,"color":"k"},label="Median")
    plt.legend()
```
