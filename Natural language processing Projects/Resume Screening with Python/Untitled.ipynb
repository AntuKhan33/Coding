```python
import re

def cleanResume(txt):
    """
    Cleans the resume text by removing unwanted characters, URLs, special characters, 
    and excessive whitespace to make it suitable for text analysis.
    """
    # Remove URLs (e.g., "http://example.com" or "https://example.com") and trailing spaces after them
    cleanText = re.sub(r'http\S+\s', ' ', txt)
    
    # Remove "RT" (used for retweets) and "cc" (common shorthand for carbon copy)
    cleanText = re.sub(r'\b(RT|cc)\b', ' ', cleanText)
    
    # Remove hashtags (e.g., "#example") but keep the text without the hashtag
    cleanText = re.sub(r'#\S+\s', ' ', cleanText)
    
    # Remove mentions (e.g., "@username")
    cleanText = re.sub(r'@\S+', ' ', cleanText)
    
    # Remove punctuation and special characters using re.escape to handle all safely
    cleanText = re.sub(r'[%s]' % re.escape("""!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~"""), ' ', cleanText)
    
    # Remove non-ASCII characters (e.g., emojis, foreign language characters)
    cleanText = re.sub(r'[^\x00-\x7F]', ' ', cleanText)
    
    # Replace multiple spaces (or other forms of whitespace like tabs and newlines) with a single space
    cleanText = re.sub(r'\s+', ' ', cleanText)
    
    # Strip leading and trailing whitespace for cleaner output
    cleanText = cleanText.strip()
    
    return cleanText
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
X_train= X_train.toarray() if hasattr (X_train,'toarray') else X_train
X_test = X_test.toarray() if hasattr(X_test, 'toarray') else X_test
```

```python
knn_model=OneVsRestClassifier(KNeighborsClassifier())
knn_model.fit(X_train,y_train)
y_pred_knn=knn_model.predict(X_test)
```

```python
print("\n KneighborsClassifier Result:")
print(f"Accuracy: {accuracy_score(y_test,y_pred_knn):.4f}")
print(f"confusion Matrix: {confusion_matrix(y_test,y_pred_knn)}")
print(f"Classification Report{classification_report(y_test,y_pred_knn)}")
```

```python
# 2. Train SVC
svc_model = OneVsRestClassifier(SVC())
svc_model.fit(X_train, y_train)
y_pred_svc = svc_model.predict(X_test)
print("\nSVC Results:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_svc):.4f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_pred_svc)}")
print(f"Classification Report:\n{classification_report(y_test, y_pred_svc)}")
```

```python
rf_model = OneVsRestClassifier(RandomForestClassifier())
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)
print("\nRandomForestClassifier Results:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_pred_rf)}")
print(f"Classification Report:\n{classification_report(y_test, y_pred_rf)}")
```
