```python
import tensorflow as tf
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

sentences = [
    'আমি বই ভালবাসি',
    'আমি বই পড়তে ভালবাসি!',
    ]

tokenizer = Tokenizer(num_words = 10)
```

```python
tokenizer.num_words
```

```python

tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
print(word_index)
```

```python
tokenizer.word_counts
```
