```python

from tensorflow.keras.preprocessing.text import Tokenizer

sentences = [
    'আমি ভালবাসি বই পড়তে।'
        ]

tokenizer = Tokenizer(num_words = 10, filters='!।')
```

```python

tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
print(word_index)
```

```python

tokenizer.num_words
```
